## 3.1 将python作为计算器

-+*/

想要进行向下取整得到整数结果，可以使用 // 运算符

计算余数，则可以使用 % 运算符

可以利用 ** 运算符来计算幂

操作数类型不同时，运算符会自动将整数操作数转换为浮点数

在交互模式下，最后输出的数值会被赋值给变量 _(不能给_赋值)

Python 还内置了对复数的支持，使用 j 或 J 作为后缀来表示虚数部分（例如 3+5j ）
