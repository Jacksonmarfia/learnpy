字符串可以被赋予索引值，第一个字符的索引值为 0。字符串并没有独立的字符类型；实际上，每个字符都可以被视为长度为 1 的字符串而已。

```python
# 前面包含后面不包含
word = "Python"
print(word[0])  # Prints 'P'
print(word[-1])  # Prints 'n'
print(word[-2])  # Prints 'o'
print(word[1:4])  # Prints 'yth'
print(word[1:4])  # Prints 'yth'
print(word[:2])  # Prints 'Py'
print(word[2:4])  # Prints 'thon'
print(word[-2:])  # Prints 'on'

print(word[:2]+word[2:])  # Prints 'Python'
```

Python 中的字符串是不可更改的——它们是不可变的。因此，试图将某个值赋给字符串中的某个索引位置会引发错误

## 3.3 列表

```python
squares = [1, 4, 9, 16, 25]

print(squares + [36, 49, 64, 81, 100])  # Prints 1
```

与不可变的字符串不同，列表是一种可变类型，也就是说，可以修改其内容。

```python
print(squares)  
squares[3] = 15
print(squares) 
```

你还可以使用 list.append() 方法在列表末尾添加新项

```python
squares.append(36)
print(squares)
```

简单的赋值操作不会复制数据。当你把一个列表赋值给某个变量时，该变量实际上只是指向了原来的列表。
你对这个列表所做的任何修改，都会同时反映在所有引用该列表的变量中。

```python
rgb = ["Red", "Green", "Blue"]
rgba = rgb
print(id(rgba) == id(rgb))  # Returns True

rgba.append("Alph")
print(rgb)  # Prints ['Red', 'Green', 'Blue', 'Alpha']
```

```python
# 复制和引用区别

cor_rgba = rgba[:]
print(id(cor_rgba) == id(rgba))
cor_rgba = rgba
print(id(cor_rgba) == id(rgba))

cor_rgba[-1] = "alpha"
print(cor_rgba)
print(rgba)  
```

```python
letters = ['a', 'b', 'c', 'd', 'e', 'f', 'g']
letters[0] = 'b'
print(letters[0]) 

letters[2:5] = ['C', 'D', 'E']
print(letters) 

letters = []
print(letters) 
print(len(letters))
```

```python
a = [1,2,3]
b = ['a', 'b', 'c']
c = [a, b]

print(c[0])
print(c[0][1])
```

## 3.4 程序入门

先计算右侧，再做赋值

```python
a,b = 0,1
while a < 10:
    print(a, end=',')
    a, b = b, a+b    
print('\n')

a,b = 0,1
while a < 10:
    print(a, end=',')
    c = a+b
    a = b
    b = c
```
