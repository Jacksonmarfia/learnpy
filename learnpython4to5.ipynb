## 4.1 if语句

可以有零个或多个 elif 部分，而 else 部分则是可选的。关键字‘ elif ’其实是‘else if’的缩写，它的存在有助于避免不必要的缩进。

```python
x = int(input("Enter a number: "))

if x < 0:
    print("Negative number")
elif x == 0:
    print("Zero")
else:
    print("Positive number")
```

## 4.2 for语句

Python 的 for 语句会按照序列中元素的原始顺序来遍历它们

```python
words = ['cat', 'window', 'defenestrate']
for w in words:
    print(w, len(w))
```

在遍历集合的同时修改该集合的代码往往比较难以实现。相比之下，直接遍历集合的副本或创建一个新的集合会更为简单明了。

```python
# Create a sample collection
users = {'Hans': 'active', 'Éléonore': 'inactive', '景太郎': 'active'}

# Strategy:  Iterate over a copy
for user, status in users.copy().items():
    if status == 'inactive':
        del users[user]

for user, status in users.copy().items():
    print(users[user])

print(users)

# Strategy:  Create a new collection
active_users = {}
for user, status in users.items():
    if status == 'active':
        active_users[user] = status

print(users)
```

## 4.3 range()函数

如果你确实需要遍历一系列数字，那么内置函数 range() 会非常有用。该函数可以用来生成等差数列。

```python
for i in range(5):
    print(i)
print(range(0, 5))
print(list(range(5)))
```

给定的终点永远不会成为生成序列的一部分
range(10) 会生成 10 个数值，这些数值正好对应长度为 10 的序列中的各个位置。
也可以让数值范围的起始点有所不同，或者指定不同的增量值（甚至是负数；有时这被称为“步长”）。

```python
print(list(range(5, 10)))
print(list(range(0, 10, 3)))
print(list(range(-10, -100, -30)))
```

要遍历序列中的各个索引，可以将 range() 和 len() 组合起来使用

```python
a = ['Mary', 'had', 'a', 'little', 'lamb']
for i in range(len(a)):
    print(i, a[i])

# 大多数情况下用enumerate()更方便些
names = ["Alice", "Bob", "Charlie"]
for index, name in enumerate(names):
    print(index, name)
for index, name in enumerate(names, start = 1):
    print(index, name)
```

 range() 返回的对象表现得就像是一个列表，但实际上并非如此。它是一种在迭代时能够依次返回序列中各个元素的对象。不过，它并不真正创建一个列表，从而节省了空间。

```python
print(range(10))
print(sum(range(10)))
print(list(range(5, 10)))
```

## 4.4 break和continue语句

break 语句用于跳出最内层的 for 或 while 循环。

continue 语句会继续执行循环的下一次迭代。

```python
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0:
            print(f"{n} equals {x} * {n//x}")
            break
```

```python
for num in range(2, 10):
    if num % 2 == 0:
        print(f"Found an even number {num}")
        continue
    print(f"Found an odd number {num}")
```

## 4.5 else关于循环规则

在 for 或 while 循环中， break 语句可以与 else 子句配对使用。如果循环在未执行 break 的情况下结束，则会执行 else 子句。

在 for 循环中， else 语句会在循环完成最后一次迭代之后才被执行，也就是说，只有在没有发生“break”操作的情况下才会执行该语句。

如果循环是由于 break 而提前终止的，那么 else 语句就不会被执行。当然，其他导致循环提前终止的方式，比如 return 或抛出异常，也会使得 else 语句不被执行。

```python
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0:
            print(f"{n} equals {x} * {n//x}")
            break
    else:
        # loop fell through without finding a factor
        print(f"{n} is a prime number")

# 条件始终为假才会执行else语句块。
```

try 语句中的 else 子句会在没有异常发生时执行；同样，循环中的 else 子句则会在没有 break 情况发生时执行。

## 4.7 match语句

match 语句接收一个表达式，将其值与一个或多个 case 块中指定的模式进行比较。

只有第一个匹配成功的模式会被执行，此外，该语句还可以从匹配的结果中提取出各个组成部分（如序列中的元素或对象的属性），并将其存储在变量中。

```python
# def http_error(status):
#     match status:
#         case 400:
#             return "Bad request"
#         case 404:
#             return "Not found"
#         case 418:
#             return "I'm a teapot"
#         case _:
#             return "Something's wrong with the internet"
```

变量名 _ 相当于通配符，必然能够匹配到相应的内容。如果没有任何匹配项，那么所有的分支都不会被执行。

case 401 | 403 | 404:

    return "Not allowed"

用|进行组合,名为亦或.

也可以用于给变量赋值

point is an (x, y) tuple

match point:

    case (0, 0):
        print("Origin")

    case (0, y):
        print(f"Y={y}")

    case (x, 0):
        print(f"X={x}")
    
    case (x, y):
        print(f"X={x}, Y={y}")
    
    case _:
        raise ValueError("Not a point")

```python
import ast
point = input("Enter a point coordinate (x, y): ")
point = ast.literal_eval(point)

match point:
    case (0, 0):
        print("Origin")
    case (0, y):
        print(f"Y={y}")
    case (x, 0):
        print(f"X={x}")
    case (x, y):                        #能捕获两个值
        print(f"X={x}, Y={y}")
    case _:
        raise ValueError("Not a point")

# 类似于赋值后输出
```

```python
class Point:
    def __init__(self, x, y):

        self.x = x
        self.y = y

p1 = Point(1, 0)
p2 = Point(0, 0)
p3 = Point(0, "y")

def where_is(point):
    match point:
        case Point(x=0, y=0):
            print("Origin")
        case Point(x=0, y=y):          #"y"满足此条件
            print(f"Y={y}")
        case Point(x=x, y=0):
            print(f"X={x}")
        case Point():
            print("Somewhere else")
        case _:
            print("Not a point")

where_is(p1)
where_is(p2)
where_is(p3)
```

通过在类中设置 __match_args__ 这个特殊属性来为各个属性指定具体的位置。

```python
class Point:
    __match_args__ = ('x', 'y')     #只影响case Point(x, y) 这种写法

    def __init__(self, x, y):

        self.x = x
        self.y = y
def where_is(point):
    match point:
        case Point(y, x):          #修改位置
            print(f"X={x},Y={y}")
        case Point(x, y):          #修改位置
            print(f"X={x},Y={y}")
        case _:
            print("Not a point")
var = 1
p4 = Point(0, y=var)     #（0，1）
p5 = Point(y=0, x=var)   #（1，0）
p6 = Point(0, var)       #（0，1）  
where_is(p4)             #跳过case Point(y=0, x=y)
where_is(p5)
where_is(p6)
```
