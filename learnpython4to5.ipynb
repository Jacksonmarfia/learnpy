## 4.1 if语句

可以有零个或多个 elif 部分，而 else 部分则是可选的。关键字‘ elif ’其实是‘else if’的缩写，它的存在有助于避免不必要的缩进。

## 4.2 for语句

Python 的 for 语句会按照序列中元素的原始顺序来遍历它们

```python
words = ['cat', 'window', 'defenestrate']
for w in words:
    print(w, len(w))
```

在遍历集合的同时修改该集合的代码往往比较难以实现。相比之下，直接遍历集合的副本或创建一个新的集合会更为简单明了。

```python
# Create a sample collection
users = {'Hans': 'active', 'Éléonore': 'inactive', '景太郎': 'active'}

# Strategy:  Iterate over a copy
for user, status in users.copy().items():
    if status == 'inactive':
        del users[user]

for user, status in users.copy().items():
    print(users[user])

print(users)

# Strategy:  Create a new collection
active_users = {}
for user, status in users.items():
    if status == 'active':
        active_users[user] = status

print(users)
```

## 4.3 range()函数

如果你确实需要遍历一系列数字，那么内置函数 range() 会非常有用。该函数可以用来生成等差数列。

```python
for i in range(5):
    print(i)
print(range(0, 5))
print(list(range(5)))
```

给定的终点永远不会成为生成序列的一部分
range(10) 会生成 10 个数值，这些数值正好对应长度为 10 的序列中的各个位置。
也可以让数值范围的起始点有所不同，或者指定不同的增量值（甚至是负数；有时这被称为“步长”）。

```python
print(list(range(5, 10)))
print(list(range(0, 10, 3)))
print(list(range(-10, -100, -30)))
```

要遍历序列中的各个索引，可以将 range() 和 len() 组合起来使用

```python
a = ['Mary', 'had', 'a', 'little', 'lamb']
for i in range(len(a)):
    print(i, a[i])

# 大多数情况下用enumerate()更方便些
names = ["Alice", "Bob", "Charlie"]
for index, name in enumerate(names):
    print(index, name)
for index, name in enumerate(names, start = 1):
    print(index, name)
```

 range() 返回的对象表现得就像是一个列表，但实际上并非如此。它是一种在迭代时能够依次返回序列中各个元素的对象。不过，它并不真正创建一个列表，从而节省了空间。

```python
print(range(10))
print(sum(range(10)))
print(list(range(5, 10)))
```

## 4.4 break和continue语句

break 语句用于跳出最内层的 for 或 while 循环。

continue 语句会继续执行循环的下一次迭代。

```python
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0:
            print(f"{n} equals {x} * {n//x}")
            break
```

```python
for num in range(2, 10):
    if num % 2 == 0:
        print(f"Found an even number {num}")
        continue
    print(f"Found an odd number {num}")
```

## 4.5 else关于循环规则

在 for 或 while 循环中， break 语句可以与 else 子句配对使用。如果循环在未执行 break 的情况下结束，则会执行 else 子句。

在 for 循环中， else 语句会在循环完成最后一次迭代之后才被执行，也就是说，只有在没有发生“break”操作的情况下才会执行该语句。

如果循环是由于 break 而提前终止的，那么 else 语句就不会被执行。当然，其他导致循环提前终止的方式，比如 return 或抛出异常，也会使得 else 语句不被执行。

```python
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0:
            print(f"{n} equals {x} * {n//x}")
            break
    else:
        # loop fell through without finding a factor
        print(f"{n} is a prime number")

# 条件始终为假才会执行else语句块。
```

try 语句中的 else 子句会在没有异常发生时执行；同样，循环中的 else 子句则会在没有 break 情况发生时执行。

## 4.7 match语句

match 语句接收一个表达式，将其值与一个或多个 case 块中指定的模式进行比较。

只有第一个匹配成功的模式会被执行，此外，该语句还可以从匹配的结果中提取出各个组成部分（如序列中的元素或对象的属性），并将其存储在变量中。

```python
# def http_error(status):
#     match status:
#         case 400:
#             return "Bad request"
#         case 404:
#             return "Not found"
#         case 418:
#             return "I'm a teapot"
#         case _:
#             return "Something's wrong with the internet"
```

变量名 _ 相当于通配符，必然能够匹配到相应的内容。如果没有任何匹配项，那么所有的分支都不会被执行。

case 401 | 403 | 404:

    return "Not allowed"

用|进行组合,名为亦或.

也可以用于给变量赋值

point is an (x, y) tuple

match point:

    case (0, 0):
        print("Origin")

    case (0, y):
        print(f"Y={y}")

    case (x, 0):
        print(f"X={x}")
    
    case (x, y):
        print(f"X={x}, Y={y}")
    
    case _:
        raise ValueError("Not a point")

```python
import ast
point = input("Enter a point coordinate (x, y): ")
point = ast.literal_eval(point)

match point:
    case (0, 0):
        print("Origin")
    case (0, y):
        print(f"Y={y}")
    case (x, 0):
        print(f"X={x}")
    case (x, y):                        #能捕获两个值
        print(f"X={x}, Y={y}")
    case _:
        raise ValueError("Not a point")

# 类似于赋值后输出
```

```python
class Point:
    def __init__(self, x, y):

        self.x = x
        self.y = y

p1 = Point(1, 0)
p2 = Point(0, 0)
p3 = Point(0, "y")

def where_is(point):
    match point:
        case Point(x=0, y=0):
            print("Origin")
        case Point(x=0, y=y):          #"y"满足此条件
            print(f"Y={y}")
        case Point(x=x, y=0):
            print(f"X={x}")
        case Point():
            print("Somewhere else")
        case _:
            print("Not a point")

where_is(p1)
where_is(p2)
where_is(p3)
```

通过在类中设置 __match_args__ 这个特殊属性来为各个属性指定具体的位置。

```python
class Point:
    __match_args__ = ('x', 'y')     #只影响case Point(x, y) 这种写法

    def __init__(self, x, y):

        self.x = x
        self.y = y
def where_is(point):
    match point:
        case Point(y, x):          #修改位置
            print(f"X={x},Y={y}")
        case Point(x, y):          #修改位置
            print(f"X={x},Y={y}")
        case _:
            print("Not a point")
var = 1
p4 = Point(0, y=var)     #（0，1）
p5 = Point(y=0, x=var)   #（1，0）
p6 = Point(0, var)       #（0，1）  
where_is(p4)             #跳过case Point(y=0, x=y)
where_is(p5)
where_is(p6)
```

我们可以在模式中添加一个 if 子句，这被称为“守卫条件”。如果该守卫条件不成立， match 会继续尝试下一个处理块。需要注意的是，值的捕获是在守卫条件被评估之前完成的。

```python
match point:
    case Point(x, y) if x == y:
        print(f"Y=X at {x}")
    case Point(x, y):
        print(f"Not on the diagonal")
```

模式可以任意嵌套

```python
class Point:
    __match_args__ = ('x', 'y')
    def __init__(self, x, y):
        self.x = x
        self.y = y
def print_points(points):
    match points:
        case []:
            print("No points")
        case [Point(0, 0)]:
            print("The origin")
        case [Point(x, y)]:
            print(f"Single point {x}, {y}")
        case [Point(0, y1), Point(0, y2) as p2]:
            print(f"Two on the Y axis at {y1}, {y2}")
            print(f"Second point is ({p2.x}, {p2.y})")
        case _:
            print("Something else")

p1 = Point(0, 0)
p2 = Point(0, 1)
points = [p1, p2]
print_points(points)
```

## 4.8 函数的定义

函数体中的第一行代码可以是一个字符串字面量。这个字符串字面量实际上就是该函数的文档字符串。 

函数的执行会创建一个新的符号表，该符号表用于存储函数中的局部变量。

函数中的所有变量赋值操作都会将值存储在局部符号表中。

系统会先检查局部符号表，然后再检查包含该函数的更上层函数的局部符号表，接着是全局符号表，最后是内置函数名对应的符号表。

当一个函数调用另一个函数，或者递归地调用自身时，系统会为这次调用创建一个新的局部符号表。

```python
def fib(n):
    result = []
    a, b = 0, 1
    while a < n:
        result.append(a)
        a, b = b, a + b
    return result

fib(100)
```

append() 是专门为列表对象定义的

它的作用是在列表的末尾添加一个新元素。在这个例子中，该方法相当于 result = result + [a] 的功能，但效率更高。

该方法的名称为 obj.methodname ，其中 obj 指的是某个对象（可以是一个表达式），而 methodname 则是该对象类型所定义的方法名称。

## 4.9 函数定义更多内容

### 4.9.1 默认参数值

为一个或多个参数指定默认值。

例子还介绍了 in 关键字。该关键字用于检测某个序列中是否包含特定的值。

默认值是在函数定义所在的范围内进行计算的。

```python
def ask_ok(prompt, retries=4, reminder='Please try again!'):
    while True:
        reply = input(prompt)
        if reply in {'y', 'ye', 'yes'}:
            return True
        if reply in {'n', 'no', 'nop', 'nope'}:
            return False
        retries = retries - 1
        if retries < 0:
            raise ValueError('invalid user response')
        print(reminder)

ask_ok('Do you really want to quit?')
```

```python
i = 5

def f(arg=i):
    print(arg)

i = 6
f()
f(i)
```

```python
# 下面的函数会累积后续调用中传入的参数

def f(a, L=[]):
    L.append(a)
    return L

print(f(1))
print(f(2))

# 不希望默认值在后续的调用中被共享
def f(a, L=None):
    if L is None:
        L = []
    L.append(a)
    return L

print(f(1))
print(f(2))
```

### 4.9.2 关键字参数

在函数调用中，关键字参数必须紧跟在位置参数之后。所有传递的关键字参数都必须与函数所接受的参数之一相匹配。

不能重复定义

```python
def cheeseshop(kind, 
               *arguments, 
               **keywords):
    print("-- Do you have any", kind, "?")
    print("-- I'm sorry, we're all out of", kind)
    for arg in arguments:
        print(arg)
    print("-" * 40)
    for kw in keywords:
        print(kw, ":", keywords[kw])

# *arguments接收字典，**keywords接收字典（一一对应）
cheeseshop("Limburger", "It's very runny, sir.1",
           "It's really very, VERY runny, sir.2",
           shopkeeper="Michael Palin",
           client="John Cleese",
           sketch="Cheese Shop Sketch")
```

### 4.9.3 特殊参数

仅限按位置传递的参数会标有 / 符号。不能通过关键字来传递。

 / 符号则用于将这类参数与其他参数区分开来。

 在参数列表中、第一个“仅通过关键字传递”的参数之前，加上 * 符号。

```python
def standard_arg(arg):
    print(arg)

def pos_only_arg(arg, /, res):  # 仅与位置相关
    print(arg,res)

def kwd_only_arg(*, arg):  # 仅包含关键字
    print(arg)

def combined_example(pos_only, /, standard, *, kwd_only):
    print(pos_only, standard, kwd_only)
```

```python
standard_arg(2)
pos_only_arg(1, 2)
kwd_only_arg(arg=3)
combined_example(1, 2, kwd_only=3)
```

参数名称和键名字可能发生冲突，使用/则‘name’可作为关键字使用

```python
def foo(name, /, **kwargs):
    print(name)
    return 'name' in kwargs

foo("Alice", age=30, name="New York")
```
